```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.linear_model import LogisticRegression
from sklearn.ensemble import RandomForestClassifier
from sklearn.metrics import confusion_matrix, accuracy_score, classification_report
import pandas as pd
```

```python
seed=42
X, y = make_classification(n_samples=1000, n_features=35, n_classes=2, flip_y=0.1)
X_train, X_test, train_y, test_y= train_test_split(X, y, test_size=0.3, random_state=seed)
```

```python
lr_clf=LogisticRegression(random_state=seed, solver='lbfgs').fit(X_train, train_y)
```

```python
rf_clf=RandomForestClassifier(n_estimators=50).fit(X_train, train_y)
```

```python
models={
    "Logistic Regression" : lr_clf,
    "Random Forest" : rf_clf
}
```

```python
results=[]
for name, model in models.items():
    y_pred=model.predict(X_test) 
    acc=round(accuracy_score(test_y, y_pred), 3) 
    results.append({"Model": name, "Accuracy" : acc})
```

```python
df_results=pd.DataFrame(results)
print(results)
```

```python
for name, model in models.items():
    y_pred=model.predict(X_test)
    print(f"\n===={name}====")
    print("Confusion Matrix:\n", confusion_matrix(test_y, y_pred))
    print("Classification Report:\n", classification_report(test_y, y_pred)) 
```
